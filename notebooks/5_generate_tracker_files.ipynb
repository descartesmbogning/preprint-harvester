# Build the 3 core files for preprint tracker

```python
from __future__ import annotations

from pathlib import Path
from typing import Optional, Tuple, Dict
import pandas as pd
import numpy as np


def generate_tracker_files_from_dedupe_v2(
    df: pd.DataFrame,
    out_dir: str | Path = "outputs/tracker_data",
    *,
    # core cols
    server_col: str = "server_name",
    record_id_col: str = "record_id",
    group_id_col: str = "dup_group_id",
    hierarchy_col: str = "records_hierarchy",
    date_col: str = "date_first_seen",
    year_col: str = "publication_year_first_seen",
    has_published_col: str = "has_published_version",
    source_col: Optional[str] = None,

    # outputs
    write_csv: bool = True,
    write_xlsx: bool = False,
    return_frames: bool = True,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, pd.DataFrame, Dict]:
    """
    Build tracker files from a deduped long table.

    Outputs:
      1) summary.csv
      2) yearly.csv                  -> wide, compatible with current app
      3) yearly_long_enriched.csv    -> new enriched yearly table
      4) clusters_summary.csv
      5) build_metrics.json

    Important modeling choices:
      - n_records: raw harvested row count
      - n_unique: unique dup_group_id count per server participation
      - n_is_version_of: unique clusters with cluster_size > 1 per server participation
      - n_published: unique clusters with any published version per server participation
      - yearly counts: computed from *parent rows only* to avoid double counting versions
    """
    out_dir = Path(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    required = [server_col, record_id_col, group_id_col]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    work = df.copy()

    # ---------------------------------------------------------
    # 0) Clean basics
    # ---------------------------------------------------------
    work[server_col] = work[server_col].astype("string").fillna("").str.strip()
    work[group_id_col] = work[group_id_col].astype("string").fillna("").str.strip()
    work[record_id_col] = work[record_id_col].astype("string").fillna("").str.strip()

    work = work[(work[server_col] != "") & (work[group_id_col] != "")].copy()

    # normalize published flag
    if has_published_col in work.columns:
        hp_raw = work[has_published_col]

        hp_norm = (
            hp_raw.astype("string")
            .str.strip()
            .str.lower()
            .map({
                "true": True, "false": False,
                "1": True, "0": False,
                "yes": True, "no": False,
                "y": True, "n": False
            })
        )

        # fallback for already boolean/numeric columns
        if hp_norm.notna().sum() == 0:
            hp_num = pd.to_numeric(hp_raw, errors="coerce")
            hp_norm = hp_num.map({1: True, 0: False})

        work[has_published_col] = hp_norm.fillna(False).astype(bool)
    else:
        work[has_published_col] = False

    # parse date
    if date_col in work.columns:
        work[date_col] = pd.to_datetime(work[date_col], errors="coerce")
    else:
        work[date_col] = pd.NaT

    # parse year
    if year_col in work.columns:
        y = pd.to_numeric(work[year_col], errors="coerce")
        work[year_col] = y.where((y >= 1000) & (y <= 3000)).round().astype("Int64")
    else:
        work[year_col] = pd.NA

    # ---------------------------------------------------------
    # 1) Build cluster-level summaries
    # ---------------------------------------------------------
    cluster_size = work.groupby(group_id_col, as_index=True).size().rename("cluster_size")
    n_servers = work.groupby(group_id_col)[server_col].nunique().rename("n_servers")
    cluster_has_published = work.groupby(group_id_col)[has_published_col].any().rename("cluster_has_published")

    # determine cluster parent
    if hierarchy_col in work.columns:
        h = work[hierarchy_col].astype("string").fillna("").str.lower().str.strip()
        is_parent = h.eq("parent")
    else:
        is_parent = pd.Series(False, index=work.index)

    parents = work[is_parent].copy()

    if parents.empty:
        tmp = work.sort_values(
            [group_id_col, date_col, record_id_col],
            ascending=[True, True, True],
            na_position="last"
        )
        parents = tmp.groupby(group_id_col, sort=False).head(1).copy()
    else:
        parents = parents.sort_values(
            [group_id_col, date_col, record_id_col],
            ascending=[True, True, True],
            na_position="last"
        )
        parents = parents.groupby(group_id_col, sort=False).head(1).copy()

    clusters_summary = (
        parents[[group_id_col, record_id_col, server_col, date_col, year_col]]
        .rename(columns={
            record_id_col: "parent_record_id",
            server_col: "parent_server_name",
            date_col: "parent_date_first_seen",
            year_col: "parent_year_first_seen",
        })
        .set_index(group_id_col)
        .join(cluster_size)
        .join(n_servers)
        .join(cluster_has_published)
        .reset_index()
    )

    clusters_summary["has_versions"] = clusters_summary["cluster_size"] > 1
    clusters_summary["is_cross_server"] = clusters_summary["n_servers"] > 1

    # ---------------------------------------------------------
    # 2) Summary per server
    # ---------------------------------------------------------
    n_records = work.groupby(server_col)[record_id_col].count().rename("n_records")
    n_unique = work.groupby(server_col)[group_id_col].nunique().rename("n_unique")

    version_clusters = set(
        clusters_summary.loc[clusters_summary["has_versions"], group_id_col].astype(str)
    )
    published_clusters = set(
        clusters_summary.loc[clusters_summary["cluster_has_published"], group_id_col].astype(str)
    )

    n_is_version_of = (
        work[work[group_id_col].isin(version_clusters)]
        .groupby(server_col)[group_id_col].nunique()
        .rename("n_is_version_of")
    )

    n_published = (
        work[work[group_id_col].isin(published_clusters)]
        .groupby(server_col)[group_id_col].nunique()
        .rename("n_published")
    )

    summary = (
        pd.concat([n_records, n_unique, n_is_version_of, n_published], axis=1)
        .fillna(0)
        .reset_index()
        .rename(columns={server_col: "server_name"})
    )

    for c in ["n_records", "n_unique", "n_is_version_of", "n_published"]:
        summary[c] = pd.to_numeric(summary[c], errors="coerce").fillna(0).astype(int)

    summary["pct_published"] = np.where(
        summary["n_unique"] > 0,
        (summary["n_published"] / summary["n_unique"]) * 100.0,
        np.nan,
    )

    if source_col and source_col in df.columns:
        tmp = df[[server_col, source_col]].copy()
        tmp[server_col] = tmp[server_col].astype("string").fillna("").str.strip()
        tmp[source_col] = tmp[source_col].astype("string").fillna("").str.strip()

        mode_source = (
            tmp.groupby(server_col)[source_col]
            .agg(lambda x: x.value_counts().index[0] if len(x.dropna()) else "")
            .rename("source")
            .reset_index()
            .rename(columns={server_col: "server_name"})
        )
        summary = summary.merge(mode_source, on="server_name", how="left")

    summary["collection_date"] = pd.Timestamp.utcnow().date().isoformat()

    # ---------------------------------------------------------
    # 3) Enriched yearly_long from parent clusters only
    # ---------------------------------------------------------
    parents_for_yearly = clusters_summary.copy()
    parents_for_yearly = parents_for_yearly.dropna(subset=["parent_year_first_seen"]).copy()
    parents_for_yearly["parent_year_first_seen"] = parents_for_yearly["parent_year_first_seen"].astype("Int64")

    # add numeric yearly aggregators
    parents_for_yearly["count_preprints"] = 1
    parents_for_yearly["count_unique_clusters"] = 1
    parents_for_yearly["count_versioned"] = parents_for_yearly["has_versions"].fillna(False).astype(int)
    parents_for_yearly["count_published"] = parents_for_yearly["cluster_has_published"].fillna(False).astype(int)
    parents_for_yearly["count_cross_server"] = parents_for_yearly["is_cross_server"].fillna(False).astype(int)

    parents_for_yearly["count_single_version"] = (~parents_for_yearly["has_versions"].fillna(False)).astype(int)
    parents_for_yearly["count_not_published"] = (~parents_for_yearly["cluster_has_published"].fillna(False)).astype(int)
    parents_for_yearly["count_not_cross_server"] = (~parents_for_yearly["is_cross_server"].fillna(False)).astype(int)

    yearly_long_enriched = (
        parents_for_yearly
        .groupby(["parent_server_name", "parent_year_first_seen"], as_index=False)
        .agg(
            count_preprints=("count_preprints", "sum"),
            count_unique_clusters=("count_unique_clusters", "sum"),
            count_versioned=("count_versioned", "sum"),
            count_published=("count_published", "sum"),
            count_cross_server=("count_cross_server", "sum"),
            count_single_version=("count_single_version", "sum"),
            count_not_published=("count_not_published", "sum"),
            count_not_cross_server=("count_not_cross_server", "sum"),
        )
        .rename(columns={
            "parent_server_name": "server_name",
            "parent_year_first_seen": "year",
        })
        .sort_values(["server_name", "year"])
        .reset_index(drop=True)
    )

    # ---------------------------------------------------------
    # 4) Wide yearly file for backward compatibility
    #     uses count_preprints only
    # ---------------------------------------------------------
    yearly_wide = (
        yearly_long_enriched
        .pivot_table(
            index="server_name",
            columns="year",
            values="count_preprints",
            aggfunc="sum",
            fill_value=0
        )
        .reset_index()
    )
    yearly_wide.columns = [str(c) for c in yearly_wide.columns]

    # ---------------------------------------------------------
    # 5) Add quick recent indicators into summary
    # ---------------------------------------------------------
    def _merge_year_metric(summary_df: pd.DataFrame, yl: pd.DataFrame, metric_col: str, prefix: str):
        tmp = (
            yl.pivot_table(
                index="server_name",
                columns="year",
                values=metric_col,
                aggfunc="sum",
                fill_value=0
            )
            .reset_index()
        )
        tmp.columns = [str(c) for c in tmp.columns]

        for yr in ["2024", "2025"]:
            new_col = f"{prefix}_{yr}"
            if yr in tmp.columns:
                summary_df = summary_df.merge(
                    tmp[["server_name", yr]].rename(columns={yr: new_col}),
                    on="server_name",
                    how="left"
                )
            else:
                summary_df[new_col] = 0
        return summary_df

    summary = _merge_year_metric(summary, yearly_long_enriched, "count_preprints", "count")
    summary = _merge_year_metric(summary, yearly_long_enriched, "count_versioned", "versioned")
    summary = _merge_year_metric(summary, yearly_long_enriched, "count_published", "published")

    for c in [
        "count_2024", "count_2025",
        "versioned_2024", "versioned_2025",
        "published_2024", "published_2025"
    ]:
        if c in summary.columns:
            summary[c] = pd.to_numeric(summary[c], errors="coerce").fillna(0).astype(int)

    # ---------------------------------------------------------
    # 6) Metrics / QA
    # ---------------------------------------------------------
    metrics = {
        "rows_in_input": int(len(df)),
        "rows_used": int(len(work)),
        "n_servers": int(work[server_col].nunique()),
        "n_clusters": int(clusters_summary[group_id_col].nunique()),
        "clusters_with_versions": int(clusters_summary["has_versions"].sum()),
        "clusters_with_published": int(clusters_summary["cluster_has_published"].sum()),
        "cross_server_clusters": int(clusters_summary["is_cross_server"].sum()),
        "year_min": int(yearly_long_enriched["year"].min()) if len(yearly_long_enriched) else None,
        "year_max": int(yearly_long_enriched["year"].max()) if len(yearly_long_enriched) else None,
        "out_dir": str(out_dir),
    }

    # ---------------------------------------------------------
    # 7) Write files
    # ---------------------------------------------------------
    if write_csv:
        summary.to_csv(out_dir / "summary.csv", index=False)
        yearly_wide.to_csv(out_dir / "yearly.csv", index=False)  # compatible with current app
        yearly_long_enriched.to_csv(out_dir / "yearly_long_enriched.csv", index=False)
        clusters_summary.to_csv(out_dir / "clusters_summary.csv", index=False)
        pd.DataFrame([metrics]).to_json(out_dir / "build_metrics.json", orient="records", indent=2)

    if write_xlsx:
        with pd.ExcelWriter(out_dir / "tracker_outputs.xlsx", engine="openpyxl") as xw:
            summary.to_excel(xw, sheet_name="summary", index=False)
            yearly_wide.to_excel(xw, sheet_name="yearly_wide", index=False)
            yearly_long_enriched.to_excel(xw, sheet_name="yearly_long_enriched", index=False)
            clusters_summary.to_excel(xw, sheet_name="clusters_summary", index=False)
            pd.DataFrame([metrics]).to_excel(xw, sheet_name="metrics", index=False)

    if return_frames:
        return summary, yearly_wide, yearly_long_enriched, clusters_summary, metrics

    return summary, yearly_wide, yearly_long_enriched, clusters_summary, metrics
```

```python
summary_df, yearly_wide_df, yearly_long_enriched_df, clusters_df, build_metrics = (
    generate_tracker_files_from_dedupe_v2(
        data,
        out_dir="outputs_new/tracker_data",
        write_csv=True,
        write_xlsx=False,
    )
)

print(build_metrics)
```
